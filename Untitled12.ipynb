```python
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

response = requests.get(url)
response.raise_for_status()

data = response.json()

df = pd.json_normalize(data["items"])
```

```python
df = df.copy()

for col in df.columns:
    df[col] = df[col].apply(
        lambda x: str(x) if isinstance(x, (dict, list)) else x
    )
```

```python
import sqlite3

conn = sqlite3.connect("github_projects.db")

df.to_sql(
    "Repositories",
    conn,
    if_exists="replace",
    index=False
)

print("Database created successfully!")
```

```python
pd.read_sql_query(
    "SELECT * FROM Repositories LIMIT 5",
    conn
)
```

```python
query = """
SELECT *
FROM Repositories
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, language, stargazers_count
FROM Repositories
WHERE stargazers_count > 1000;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, language, stargazers_count, forks_count
FROM Repositories
WHERE stargazers_count > 1000
AND forks_count > 100;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, language, stargazers_count
FROM Repositories
WHERE language = 'Python'
OR language = 'Jupyter Notebook';
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, language, stargazers_count
FROM Repositories
WHERE NOT language = 'Python';
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, language, stargazers_count
FROM Repositories
ORDER BY stargazers_count DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(*) AS total_repositories
FROM Repositories;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT AVG(stargazers_count) AS average_stars
FROM Repositories;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
ORDER BY repository_count DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 1
ORDER BY repository_count DESC;
"""

pd.read_sql_query(query, conn)
```
